# Exercicio 2

```python
numA = input("Digite um número A: ")
numB = input("Digite um número B: ")

print(f"{numA} + {numB} = {int(numA + numB)}")
print(f"{numA} - {numB} = {int(numA) - int(numB)}")
print(f"{numA} * {numB} = {float(numA) * int(numB)}")
print(f"{numA} * {numB} = {float(numA) / int(numB)}")


```
